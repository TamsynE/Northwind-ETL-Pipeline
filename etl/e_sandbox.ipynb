```python
from extract import extract_all
import pandas as pd

def transform(original):
    # 1. Customers Dimension
    dim_customers = original["customers"].copy()
    # dim_customers.columns = dim_customers.columns.str.lower()
    # dim_customers = dim_customers.rename(columns={"id": "customerid"})
    dim_customers = dim_customers.drop(['phone', 'fax'])


    # 2. Products Dimension (products + categories)
    dim_products = original["products"].merge(
        original["categories"],
        how="left",
        left_on="categoryID",
        right_on="categoryID"
    )

    dim_products = dim_products.drop(['supplierID', 'quantityPerUnit', 'description', 'picture'])

    # 3. Employees Dimension
    dim_employees = original["employees"].copy()
    dim_employees = dim_employees.drop(['titleOfCourtesy', 'homePhone', 'extension', 'photo', 'notes', 'photoPath'])

    # 4. Orders Fact (orders + order_details)
    fact_orders = original["order_details"].merge(
        original["orders"],
        how="left",
        on="orderID"
    )

    fact_orders.drop(['requiredDate', 'shippedDate', 'shipVia', 'freight'])

    print("Transformation complete.")
    
    return {
        # "fact_sales": fact_sales,
        "dim_employees": dim_employees,
        "dim_customers": dim_customers,
        "dim_products": dim_products
    }
```
